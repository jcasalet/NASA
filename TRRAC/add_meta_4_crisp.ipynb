```python
import pandas as pd
workDir = '/Users/jcasalet/Desktop/RESEARCH/LIVER/DATA/HI/4_CRISP/'
outputDir = workDir + 'NORMALIZE_THEN_MERGE'
metaFile = workDir + '/metadata.csv'
```

```python
def processAndSave(workDir, exprFile, metaFile, outputExprFileBase):
    # read in csv
    df=pd.read_csv(exprFile, header=0, sep=',', index_col=0)

    # transpose
    df = df.transpose()

    # re-index to set index as column name
    df.reset_index(inplace=True)

    # rename index col to sample
    df.rename(columns={"index":"sample"}, inplace=True)

    # read in spread sheet
    ssDF = pd.read_csv(metaFile, header=0, sep=',')
    
    # create dictionary with sample id as key and concatenated column strings as value
    env_dict=dict()
    for i in range(len(ssDF)):
        key=ssDF.iloc[i]['SAMPLE ID']
        flight=ssDF.iloc[i]['Group']
        if flight == "Flight":
                fvalue=1
        else:
                fvalue=0
        value=ssDF.iloc[i]['Study'] + ':' + ssDF.iloc[i]['Dissection']  + ':' + ssDF.iloc[i]['Library prep'] + ':' + str(fvalue)
        #value=ssDF.iloc[i]['Dissection']  + ':' + str(fvalue)
        env_dict[key] = value

    # join env dictionary to data frame
    df['env'] = df['sample'].map(env_dict)

    # create oro threshold dictionary with arbitrary threshold
    # >18 = 0 which is the 70% oro value, and there are 70% gc and 30% flight
    threshold=18
    orothresh_dict=dict()
    for i in range(len(ssDF)):
        key=ssDF.iloc[i]['SAMPLE ID']
        oro=ssDF.iloc[i]['ORO Positivity (%)']
        if oro < threshold:
                orothresh_dict[key] = 0
        else:
                orothresh_dict[key] = 1

    # join oro threshold dictionary to data frame
    df['oro_thresh'] = df['sample'].map(orothresh_dict)

    df = df.rename(columns = {'index':'sample'})
    
    df.to_csv(outputExprFileBase + '.csv', sep=',', index=None)
    
    df.to_pickle(outputExprFileBase + '.pkl')
```

```python
exprFile = outputDir + '/normalized_merged_corrected_log_complete.csv'
outputExprFileBase = outputDir + '/normalized_merged_corrected_log_complete_crisp'
processAndSave(workDir, exprFile, metaFile, outputExprFileBase)

exprFile = outputDir + '/normalized_merged_corrected_log_no_pseudogenes.csv'
outputExprFileBase = outputDir + '/normalized_merged_corrected_log_no_pseudogenes_crisp'
processAndSave(workDir, exprFile, metaFile, outputExprFileBase)

exprFile = outputDir + '/normalized_merged_corrected_log_no_pseudogenes_wExternal.csv'
outputExprFileBase = outputDir + '/normalized_merged_corrected_log_no_pseudogenes_wExternal_crisp'
processAndSave(workDir, exprFile, metaFile, outputExprFileBase)
```
